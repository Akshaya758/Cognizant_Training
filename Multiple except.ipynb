```python
try:
    num=int(input("Enter the number"))
    print(10/num)
except ZeroDivisionError:
    print("Only zero")
except ValueError:
    print("only Integers")
```

```python
try:
    num=complex(input("Enter the number"))
    print(10/num)
except ZeroDivisionError:
    print("Only zero")
except ValueError:
    print("only Decimals")
```

```python
try:
    num=float(input("Enter the number"))
    print(10/num)
except ZeroDivisionError:
    print("Only zero")
except ValueError:
    print("only numbers")
```
